05. (15 pontos) Deque Sequencial
Na implementação de deque sequencial feita nas aulas, criamos métodos para
inserção. Considere que a classe Deque contém apenas o atributo _itens (uma lista
Python). Escreva um novo método chamado remover(self, lado) que servirá para
remover de qualquer um dos lados, onde o parâmetro lado representa o lado a ser
removido (0 para o início ou 1 para o final). O método deve retornar o valor removido
(ou None se estiver vazio).
Obs: Não utilize métodos prontos de remoção do Deque, construa a lógica acessando
o atributo _itens.


```python
class Deque:
    def __init__(self):
        self.itens = []
    def inserir_final(self, dado):
        self.itens.append(dado)
        return True
    def remover(self, lado:int):
        if lado == 0:
            removido = self.itens[0] # Remove do início
            for i in range(len(self.itens)-1):
                self.itens[i]= self.itens[i+1]
            del self.itens[-1]
            return removido
        elif lado == 1:
            removido = self.itens[-1] # Remove do final
            del self.itens[-1]
            return removido
        return False

if __name__ == "__main__":
    fila = Deque()
    fila.inserir_final(1)    
    fila.inserir_final(2)    
    fila.inserir_final(3)    
    fila.inserir_final(4)    
    fila.inserir_final(5)

    fila.remover(0)
    fila.remover(1)

    for item in fila.itens:
        print(item)    
```

06. (15 pontos) Lista Encadeada
Considere a implementação de lista encadeada utilizada nas aulas, sendo que a classe
No contém os atributos dado e prox e a classe Lista contém apenas o atributo _inicio.
Escreva um novo método para a classe Lista chamado maior_valor(self) que retorne o
maior valor presente na lista.
Observações:
● Considere que a lista contém apenas valores numéricos.
● Se a lista estiver vazia, deve ser retornado None.


```python
class No:
    def __init__(self, dado):
        self.dado = dado
        self.prox = None
class Lista:
    def __init__(self):
        self.inicio = None
    def vazia(self):
        return self.inicio == None
    def inserir(self, dado):
        if self.vazia():
            self.inicio = No(dado)
            return True
        novo = No(dado)
        novo.prox = self.inicio
        self.inicio = novo
    def maior_valor(self):
        if self.vazia():
            return None
        atual = self.inicio
        maior = int(atual.dado)
        while atual:
            if int(atual.dado) > maior:
                maior = int(atual.dado)
            atual = atual.prox
        return maior

if __name__ == "__main__":
    lista = Lista()
    lista.inserir(1)
    lista.inserir(2)
    lista.inserir(3)
    lista.inserir(8)
    lista.inserir(4)
    lista.inserir(5)
    lista.inserir(6)
    print(lista.maior_valor())
    
    
```

Escreva uma função (fora da classe) chamada inverter_pilha(pilha) que receba como
parâmetro uma Pilha e inverta a ordem dos elementos dentro dela.
Dica: Use uma fila como estrutura auxiliar. A função não precisa ter retorno, a própria
pilha recebida deve ser modificada.

```python
from PilhaEncadeada import PilhaEncadeada
from FilaEncadeada import FilaEncadeada

def inverterPilha(pilha:PilhaEncadeada):
    if pilha.vazia():
        return None
    filaAux = FilaEncadeada()
    while not pilha.vazia():
        filaAux.enfileirar(pilha.desempilhar())
    while not filaAux.vazia():
        pilha.empilhar(filaAux.desenfileirar())
```
