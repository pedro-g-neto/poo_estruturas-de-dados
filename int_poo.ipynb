```python
class Ponto:
  def __init__(self, x, y):
    self.x = x
    self.y = y
  def quadrante(self):
    if self.x > 0 and self.y > 0:
      return 1
    elif self.x < 0 and self.y > 0:
      return 2
    elif self.x < 0 and self.y < 0:
      return 3
    elif self.x > 0 and self.y < 0:
      return 4

p1 = Ponto(1,2)
print(f"Ponto 1: x = {p1.x}; y = {p1.y}")
p2 = Ponto(5,5)
print(f"Ponto 2: x = {p2.x}; y = {p2.y}")

p2.x = 4
print(f"Ponto 2: x = {p2.x}; y = {p2.y}")

print(f"O Ponto 1 está no {p1.quadrante()}º Quadrante")
```

```python
class Retangulo:
  def __init__(self, base, altura):
    self.base = base
    self.altura = altura
  def calcular_area(self):
    return self.base*self.altura
  def eh_quadrado(self):
    return self.base == self.altura

retangulo1 = Retangulo(4,5)
retangulo2 = Retangulo(5,5)

if retangulo1.eh_quadrado():
  print(f"R1 é um quadrado")
else:
  print(f"R1 não é um quadrado")

if retangulo2.eh_quadrado():
  print(f"R2 é um quadrado")
else:
  print(f"R2 não é um quadrado")

areaR1 = retangulo1.calcular_area()
areaR2 = retangulo2.calcular_area()

print(f"A área do R1 é {areaR1} u.a; A área do R2 é {areaR2} u.a")
```

```python
class TV:
  def __init__(self):
    self.ligada = False

  def liga_desliga(self):
    self.ligada = not self.ligada
    if self.ligada:
      self.canal = 1
      self.volume = 50
  def mudar_canal(self, canal):
    self.canal = canal
  def aumentar_volume(self):
    self.volume += 1
  def diminuir_volume(self):
    self.volume -= 1
  def imprimir(self):
    print(f'Canal: {self.canal}; Volume: {self.volume}')

tv = TV()
tv.liga_desliga()
tv.imprimir()
tv.mudar_canal(5)
tv.aumentar_volume()
tv.imprimir()
```

```python
class Ponto:
  def __init__(self, x, y):
    self.__x = x
    self.__y = y

  def get_x(self):
    return self.__x

  def set_x(self,x):
    self.__x = x

  def get_y(self):
    return self.__y

  def set_y(self,y):
    self.__y = y
p = Ponto(1,2)
print(f'Ponto 1: {p.get_x()}, {p.get_y()}')
p.set_x(4)
p.set_y(5)
print(f'Ponto 1: {p.get_x()}, {p.get_y()}')
```

```python
class Ponto:
  def __init__(self, x, y):
    self.__x = x
    self.__y = y

  @property
  def x(self):
    return self.__x
  @x.setter
  def set_x(self,x):
    self.__x = x
  @property
  def y(self):
    return self.__y
  @y.setter
  def set_y(self,y):
    self.__y = y
```
