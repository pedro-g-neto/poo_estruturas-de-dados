### Introdução à Orientação a Objetos
#### Definindo uma classe simples

A classe é o molde de um objeto. Define as regras de existência, os métodos (funções) e não existe de forma física no sistema.

O objeto em si é uma instância da classe. Uma entidade que segue os moldes da classe e, portanto, possui atributos e métodos compatíveis com o projeto da classe.

```python
class Ponto:
  def __init__(self, x, y): # construtor
    self.x = x # atributos do objeto
    self.y = y
  def quadrante(self): # o self é uma referência (endereço de memória) ao objeto que está chamando o método
    if self.x > 0 and self.y > 0:
      return 1
    elif self.x < 0 and self.y > 0:
      return 2
    elif self.x < 0 and self.y < 0:
      return 3
    elif self.x > 0 and self.y < 0:
      return 4

p1 = Ponto(1,2) # Instanciando um objeto da classe Ponto
print(f"Ponto 1: x = {p1.x}; y = {p1.y}")
p2 = Ponto(5,5)
print(f"Ponto 2: x = {p2.x}; y = {p2.y}")

p2.x = 4
print(f"Ponto 2: x = {p2.x}; y = {p2.y}")

print(f"O Ponto 1 está no {p1.quadrante()}º Quadrante")
```

```python
class Retangulo:
  def __init__(self, base, altura):
    self.base = base
    self.altura = altura
  def calcular_area(self): # cada função abaixo do construtor é um método utilizável da classe
    return self.base*self.altura
  def eh_quadrado(self):
    return self.base == self.altura

retangulo1 = Retangulo(4,5)
retangulo2 = Retangulo(5,5)

if retangulo1.eh_quadrado():
  print(f"R1 é um quadrado")
else:
  print(f"R1 não é um quadrado")

if retangulo2.eh_quadrado():
  print(f"R2 é um quadrado")
else:
  print(f"R2 não é um quadrado")

areaR1 = retangulo1.calcular_area()
areaR2 = retangulo2.calcular_area()

print(f"A área do R1 é {areaR1} u.a; A área do R2 é {areaR2} u.a")
```

#### Encapsulamento

Impede o acesso direto às propriedades dos objetos, aumentando o nível de abstração.

Para acessar objetos encapsulados, é necessário que o desenvolvedor da Classe tenha definido métodos acessadores (getters) ou métodos modificadores (setters) através de métodos crus ou o uso de propriedades (decorators)

```python
class ContaCorrente:
  def __init__(self, titular, numero, saldo = 0):
    self.__numero = numero # duplo underscore para indicar que o atributo é privado
    self.__titular = titular
    self.__saldo = saldo

  def get_titular(self): # método acessador (getter) para atributo privado
    return self.__titular
  def get_saldo(self):
    return self.__saldo
  def get_numero(self):
    return self.__numero

  def set_titular(self,titular): # método modificador (setter) para atributo privado
    self.__titular = titular
  def set_numero(self,numero):
    self.__numero = numero
  def set_saldo(self, saldo):
    self.__saldo = saldo
  
  def sacar(self,valor):
    if valor <=0:
      print("Valor inválido")
    elif valor > self.__saldo:
      print("Saldo insuficiente")
    else:
      self.__saldo -= valor
      print(f"Saque de R${valor:.2f} efetuado com sucesso")

  def depositar(self,valor):
    if valor > 0:
      self.__saldo += valor
      print(f"Depósito efetuado com sucesso no valor de R${valor:.2f}")
    else:
      print("Valor inválido")

  def imprimir(self):
    print(f"Nº: {self.__numero}")
    print(f"Titular: {self.__titular}")
    print(f"Saldo: {self.__saldo}")


conta1 = ContaCorrente("Pedro", 6767)
conta1.depositar(1000)
conta1.sacar(200)
conta1.imprimir()
```

```python
class TV:
  def __init__(self):
    self.ligada = False

  def liga_desliga(self):
    self.ligada = not self.ligada
    if self.ligada:
      self.canal = 1
      self.volume = 50
  def mudar_canal(self, canal):
    self.canal = canal
  def aumentar_volume(self):
    self.volume += 1
  def diminuir_volume(self):
    self.volume -= 1
  def imprimir(self):
    print(f'Canal: {self.canal}; Volume: {self.volume}')

tv = TV()
tv.liga_desliga()
tv.imprimir()
tv.mudar_canal(5)
tv.aumentar_volume()
tv.imprimir()
```

```python
class Ponto:
  def __init__(self, x, y):
    self.__x = x
    self.__y = y

  def get_x(self):
    return self.__x

  def set_x(self,x):
    self.__x = x

  def get_y(self):
    return self.__y

  def set_y(self,y):
    self.__y = y
p = Ponto(1,2)
print(f'Ponto 1: {p.get_x()}, {p.get_y()}')
p.set_x(4)
p.set_y(5)
print(f'Ponto 1: {p.get_x()}, {p.get_y()}')
```

```python
class Ponto:
  def __init__(self, x, y):
    self.__x = x
    self.__y = y

  @property # decorator para criar um método acessador (getter) para o atributo privado x
  def x(self):
    return self.__x
  @x.setter # decorator para criar um método modificador (setter) para o atributo privado x
  def set_x(self,x):
    self.__x = x
  @property
  def y(self):
    return self.__y
  @y.setter
  def set_y(self,y):
    self.__y = y
```

#### Relacionamento entre Objetos

```python
class ContaCorrente:
  def __init__(self, cliente:str, numero:int, saldo:float = 0):
    self.__numero = numero
    self.__cliente = cliente
    self.__saldo = saldo

  @property
  def cliente(self):
    return self.__cliente
  @property
  def saldo(self):
    return self.__saldo
  @property
  def numero(self):
    return self.__numero
  @cliente.setter
  def cliente(self,cliente:str):
    self.__cliente = cliente
  @numero.setter
  def numero(self,numero:int):
    self.__numero = numero
  @saldo.setter
  def saldo(self, saldo:float):
    self.__saldo = saldo
  
  def sacar(self,valor:float):
    if valor <=0:
      print("Valor inválido")
    elif valor > self.__saldo:
      print("Saldo insuficiente")
    else:
      self.__saldo -= valor
      print(f"Saque de R${valor:.2f} efetuado com sucesso")

  def depositar(self,valor:float):
    if valor > 0:
      self.__saldo += valor
      print(f"Depósito efetuado com sucesso no valor de R${valor:.2f}")
    else:
      print("Valor inválido")

  def imprimir(self):
    print(f"Nº: {self.__numero}")
    print(f"Titular: {self.__cliente}")
    print(f"Saldo: {self.__saldo}")


class Cliente:
  def __init__(self,nome:str,idade:int, cidade:str, estado:str):
    self.__nome = nome
    self.__idade = idade
    self.__cidade = cidade
    self.__estado = estado

  @property
  def nome(self):
    return self.__nome
  @property
  def idade(self):
    return self.__idade
  @property
  def cidade(self):
    return self.__cidade
  @property
  def estado(self):
    return self.__estado

  @nome.setter
  def nome(self,nome):
    self.__nome = nome
  @idade.setter
  def idade(self, idade):
    self.__idade = idade
  @cidade.setter
  def cidade(self, cidade):
    self.__cidade = cidade
  @estado.setter
  def estado(self, estado):
    self.__estado = estado

  def __str__(self):
    return f'{self.__nome}'

cliente = Cliente("Pedro", 20, "João Pessoa", "PB")
print(cliente)
conta1 = ContaCorrente(cliente, "0001", 1000)
conta1.imprimir()
print(conta1.cliente.cidade)
conta1.cliente.nome = "Pedro Gomes de Andrade Neto"
print(f'----Cliente----\nNome: {cliente}\nCidade: {cliente.cidade}-{cliente.estado}\nIdade: {cliente.idade}')
print('----Conta----')
conta1.imprimir()
```
